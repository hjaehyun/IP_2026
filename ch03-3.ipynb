```python
def saturation(value):
  if(value>255):
    value=255
  elif(value<0):
    value=0
  return value
```

```python
gOut=np.zeros((gimg.shape[0],gimg.shape[1]), dtype=np.ubyte)
```

```python
from re import M
min=256
max=0
for h in range(gimg.shape[0]):
  for w in range(gimg.shape[1]):
    if gimg[h,w] <min:
      min=gimg[h,w]
    if gimg[h,w] > max:
      max=gimg[h,w]
print(min)
print(max)
```

```python
for h in range(gimg.shape[0]):
  for w in range(gimg.shape[1]):
    val= (gimg[h,w]-min)/(max-min)*255
    gOut[h,w]=val
cv2_imshow(gOut)
```
