```python
img3 = cv2.imread('/content/gdrive/My Drive/colabIP_26/images/lena256.jpg')
RGB_img = cv2.cvtColor(img3, cv2.COLOR_BGR2RGB)
gimg3 = cv2.cvtColor(img3, cv2.COLOR_BGR2GRAY)

print(gimg1.shape)
cv2_imshow(gimg3)
```

```python
from random import randint

noiseImg = gimg3.copy()
```

```python
noiseRatio = 10

for h in range(0, gimg3.shape[0]):
  for w in range(0, gimg3.shape[1]):
    val = randint(0,100)
    if(val < noiseRatio/2):
      noiseImg[h][w] = 255
    elif(val < noiseRatio):
      noiseImg[h][w] = 0

cv2_imshow(noiseImg)
```

```python
outimg = cv2.medianBlur(noiseImg,3)
cv2_imshow(outimg)
```
