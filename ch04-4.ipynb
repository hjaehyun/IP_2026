```python
img = cv2.imread('/content/gdrive/My Drive/colabIP_26/images/lena256.jpg')
RGB_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gimg1 = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2_imshow(gimg1)
```

```python
hist = cv2.calcHist([gimg1], [0], None, [255], [0, 255])
plt.plot(hist)
```

```python
gOut = cv2.equalizeHist(gimg1)
cv2_imshow(gOut)
```

```python
hist2 =cv2.calcHist([gOut], [0], None, [255], [0, 255])
plt.plot(hist2)
```

```python
cumuHist = hist.copy()
for i in range(1, len(cumuHist)):
  cumuHist[i] = cumuHist[i-1] + hist2[i]
plt.plot(cumuHist)
#
```
