```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/gdrive')
#/content/gdrive/My Drive/colabIP_26/images/
```

```python
img = cv2.imread('/content/gdrive/My Drive/colabIP_26/images/girl.jpg')
gimg = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2_imshow(gimg)
```

```python
gPlus = np.zeros((gimg.shape[0], gimg.shape[1]), dtype=np.ubyte)
gOrigin = np.zeros((gimg.shape[0], gimg.shape[1]), dtype=np.ubyte)
```

```python
val = 120

for h in range(gimg.shape[0]):
  for w in range(gimg.shape[1]):
    pVal = np.int32(gimg[h][w]) + val
    if(pVal > 255):
      pVal = 255
    gPlus[h][w] = pVal

cv2_imshow(gPlus)
```

```python
val = 120

for h in range(gimg.shape[0]):
  for w in range(gimg.shape[1]):
    pVal = np.int32(gimg[h][w]) - val
    if(pVal > 255):
      pVal = 255
    gOrigin[h][w] = pVal

cv2_imshow(gOrigin)
```
